```python
np.random.seed(613) #random seed
x = np.arange(50) 
y = np.random.randint(0, 100,50)
```

```python
fig, ax = plt.subplots(figsize=(5, 3)) 
#fig - tell python it need to draw a figure 
#ax - axis
#figsize - 5 is width, 3 is height
ax.scatter(x,y)
#Draw an xy "scatter" plot
```

```python
fig, ax = plt.subplots(figsize=(5, 3)) 
ax.bar(x,y)
#Draw an xy "bar" plot
```

```python
fig, ax = plt.subplots(figsize=(5, 3)) 
ax.plot(x,y)
#Draw an xy line "plot"
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y, edgecolor = "black")
#no x value as it is a histogram - knowing only the frequency of y!
# we can end as ax.hist(y), but we can also add another command -- edgecolor = "black"
# to show the edge of the bars
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y) #plot can't use edge colour
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout()
```

```python
#make a dictionary with the properties we want for our labels
font1 = {'family':'sans-serif','color':'blue','size':20}
font2 = {'family':'monospace','color':'green','size':14}
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict = font1)
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
fig.tight_layout()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y, marker="*", color = "indigo") #colour but NOT colour!!!
fig.show() #optional
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y, marker='*', color = 'indigo', 
        linestyle = '--', # NEW ADDITIONS 
        linewidth = 2) # NEW ADDITIONSfig
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y,marker='*', color = "#f12666", linestyle = '--', linewidth = 2)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y,marker='*', 
        markersize = 12, # NEW ADDITIONS 
        color = '#7425b9', linestyle = '--', 
        linewidth = 2, 
        markeredgecolor = '#fa9359', # edge colour
        markerfacecolor = "#F83B3B" ) # face colour - colur inside star
```

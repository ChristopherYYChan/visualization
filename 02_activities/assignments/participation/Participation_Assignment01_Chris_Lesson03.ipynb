```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75,50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
fig,(ax1, ax2) = plt.subplots(ncols=2, #we can define (ax1,ax2) instead of ax alone like when dealing with only 1 graph
nrows=1,
figsize=(7, 3))

ax1.scatter(x1,y1) #now add the random values defined in the earlier code block
ax2.bar(x2,y2) #now add the defined values defined in the earlier code block
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
['ax2', 'ax3']], # it is a list containing two inner lists (rows), 
#and each inner list has two elements (columns), Matplotlib creates a 2x2 grid.
#ax1: Occupies the top-left cell (0, 0).
#ax2: Occupies the bottom-left cell (1, 0). 
#ax3: Because "ax3" appears in both the top-right and bottom-right positions, 
# Matplotlib merges those cells into a single tall subplot.
figsize=(7, 4))
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
['ax2', 'ax3']], # it is a list containing two inner lists (rows), 
#and each inner list has two elements (columns), Matplotlib creates a 2x2 grid.
#ax1: Occupies the top-left cell (0, 0).
#ax2: Occupies the bottom-left cell (1, 0). 
#ax3: Because "ax3" appears in both the top-right and bottom-right positions, 
# Matplotlib merges those cells into a single tall subplot.

figsize=(7, 4))
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
['ax2', 'ax3']],
figsize=(7, 4))

someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
someaxes["ax1"].set_xlabel('A Big Label',fontsize=12)
someaxes["ax2"].set_xlabel('Another Label',fontsize=12)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious',fontsize=12)
fig.tight_layout() #Now the label from ax1 can be seen
```

```python
from wordcloud import WordCloud
df = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv",
on_bad_lines='skip')
df
```

```python
# join all our text from each row from our quote column into a string
text = " ".join(each for each in df.quote)
# generate our wordcloud image
wordcloud = WordCloud(background_color="white",
colormap = 'inferno').generate(text)
# use matplotlib syntax to put our image in a figure
fig, ax = plt.subplots(figsize=(7, 3)) #size of the plot!
ax.imshow(wordcloud, # remember 'imshow' from when we added pictures to our matplotlib axes
interpolation='bilinear') # this line helps smooth our image
ax.axis("off")
```

```python
#making venn diagrams
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted
A = set(["apple", "banana", "watermelon"])
B = set(["pumpkin", "blueberry", "apple", "key lime"])
```

```python
diagram = venn2_unweighted([A, B],
set_labels = ('Fruits', 'Pies'),
set_colors=("blue", "red"),
alpha=0.5)
```

```python
diagram = venn2_unweighted([A, B],
set_labels = ('Fruits', 'Pies'),
set_colors=("blue", "red"),
alpha=0.5)
diagram.get_label_by_id("10").set_text("\n".join(A - B)) #\n new line!
diagram.get_label_by_id("11").set_text("\n".join(A & B))
diagram.get_label_by_id("01").set_text("\n".join(B - A))

```
